```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import Binarizer
```

```python
a = np.array([[1,2,3],
              [4,5,6]])
```

```python
# it use for finding dimantion
a.ndim
```

```python
# create a binarizer variable

biner = Binarizer(threshold= 3)

# convert into binery

data_new = biner.fit_transform(a)
data_new
```
