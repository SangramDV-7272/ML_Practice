```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import numpy as np
```

```python
indian = pd.read_excel("D:\\SevenMentor\\ML\\workspace\\Dataset\\Indian_Economy_Sectorwise_Dataset.xlsx")
```

```python
indian
```

```python
len(indian.Sector.unique())
```

```python
indian["GNP"] = indian["GDP_Lakh_Crore"] + indian["Imports_Crore"] - indian["Exports_Crore"]  
```

# now encoding on string data

```python
e = LabelEncoder()
```

```python
indian["Sector_en"] = e.fit_transform(indian.Sector)
```

```python
indian["Quarter_en"] = e.fit_transform(indian.Quarter)
```

```python
indian.drop(columns=["Quarter", "Sector"], inplace=True)
```

# minmax scalar in string

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
m = MinMaxScaler()
```

```python
indian.Sector_en
```

```python
indian["Sector_min_max"] = m.fit_transform(indian[["Sector_en"]])
```

```python
indian["Gdp_min_max"] = m.fit_transform(indian[["GDP_Lakh_Crore"]])
indian["Growth_%_min_max"] = m.fit_transform(indian[["Growth_%"]])
indian["GNP_%_min_max"] = m.fit_transform(indian[["GNP"]])
```

```python
indian
```

```python
indian["GNP_%_min_max"]
```

# binerization

```python
from sklearn.preprocessing import Binarizer
```

```python
bine = Binarizer(threshold=2020)
```

```python
indian["Year"] = bine.fit_transform(indian[["Year"]])
```

```python
indian
```

# Split data into train and test

```python
from sklearn.model_selection import train_test_split

X = indian.drop("Year", axis = 1) # store featuer data in "X"
y = indian["Year"] # store target in "y"
```

```python
X
```

### Split data in x_train, x_test & y_train, y_test

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=42)
```
