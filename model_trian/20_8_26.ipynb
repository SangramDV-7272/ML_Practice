```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
# import matpotlib.pypolt as plt
import matplotlib.pyplot as plt
```

```python

from sklearn.linear_model import LinearRegression
```

```python
d = pd.DataFrame({
    "quentity": [1,2,3,4,5,6],
    "price": [1000, 2000, 3000, 4000, 5000, 6000]
})
```

```python
d
```

```python
x = d[["quentity"]]
y = d["price"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, shuffle = True, random_state = 42)
```

```python
x_train
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
print("salary", model.coef_[0], "* quentity + ", model.intercept_)
```

```python
new_data = {"quentity":[[1], [2], [3]]}
```

```python
model.predict(new_data["quentity"])
```

```python
d["Predicted_price"] = model.predict(x)
print(d)
```

```python
plt.scatter(x, y)
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```

```python
plt.scatter(x, y, label="Actual Salary")
plt.plot(x, model.predict(x),label="Regression Line")
# plt.plot(, 7000,label=" Line")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.title("Linear Regression:Experience vs Salary")
plt.legend()
plt.show()
```
