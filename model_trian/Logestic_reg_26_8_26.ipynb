```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
# 1. Create DataFrame
data = {
"Study_Hours": [1, 2, 2, 3, 3, 4, 4, 5, 6, 7, 8,
9],
"Pass": [0, 0, 0, 0, 1, 0, 1, 1, 1, 1, 1, 1]
}
df = pd.DataFrame(data)
# 2. X and y
X = df[["Study_Hours"]]
y = df["Pass"]
# 3. Create model
model = LogisticRegression()
# 4. Train model
model.fit(X, y)
# 5. Predict
new_student = [[5]]
prediction = model.predict(new_student)
probability = model.predict_proba(new_student)[0]
print("Prediction:", prediction[0])
print("Fail Probability:", probability[0])
print("Pass Probability:", probability[1])
# 6. Coefficient and intercept
print("Coefficient:", model.coef_[0][0])
print("Intercept:", model.intercept_[0])
# 7. Plot Logistic Curve
X_range = np.linspace(1, 9, 100).reshape(-1, 1)
probability_curve = model.predict_proba(X_range)[:, 1]
plt.scatter(X, y, label="Actual Data")
plt.plot(X_range, probability_curve, label="LogisticCurve")
plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression")
plt.legend()
plt.show()
```
