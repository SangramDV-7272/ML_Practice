```python
import pandas as pd
from math import nan
```

```python
df = pd.DataFrame({
    "Name":["Amit", None, "Amit", "Snehal"],
    "Age":[25,25, None,28]
})
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# # fillining null value in continous data
# x = df["Age"].mean()
# df["Age"] = df["Age"].fillna(x)

# # filling null value in categrical data

# y = df["Name"].mode()[0]
# df["Name"] = df["Name"].fillna(y)
```

```python
df
```

```python
df.isna().sum()
```

```python
for i in df.columns:
    if df[i].dtype  == "str":
        x = df[i].mode()[0]
        df[i] = df[i].fillna(x)
    else:
        y = df[i].mean()
        df[i] = df[i].fillna(y)
```
