```python
X = data.drop(["Price"], axis = 1)
y = data["Price"]
```

```python
X
```

```python
cat=[]
con=[]


for i in X.columns:
    if X[i].dtype == "str":
        cat.append(i)
    else:
        con.append(i)
```

```python
categrical_columns = X[cat]
continus_columns = X[con]
```

```python
categrical_columns
```

```python
continus_columns
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output = False)
```

```python
encode = ohe.fit_transform(categrical_columns)
```

```python
encode
```

```python
ohe.get_feature_names_out()
```

```python
a = pd.DataFrame(encode, columns = ohe.get_feature_names_out().tolist())
```

```python
new_df = pd.concat([continus_columns, a], axis = 1)
```

```python
new_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(new_df, y, test_size = 0.2, random_state = 42, shuffle = True)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()

model=lr.fit(xtrain,ytrain)
```

```python
pred_tr=model.predict(xtrain)
pred_ts=model.predict(xtest)

```

```python
pred_tr
```

```python
pred_ts
```

```python
y_obse = pd.DataFrame(ytest)
```

```python
y_obse["Price_pred"] = pred_ts
```
