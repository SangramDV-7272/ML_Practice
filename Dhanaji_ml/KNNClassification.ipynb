# KNN Classification Model

```python
import pandas as pd
import numpy as np
```

# dataset 

```python
data = {
    "Studay Hours" : [1,2,2,3,3,4,5,6,7,8],
    "attendance" : [50, 55, 60, 65, 70, 72, 80, 90, 95, 97],
    "result" : ["Fail", "Fail", "Fail", "Fail", "Fail", "Pass", "Pass","Pass","Pass", "Pass"]
}
```

```python
data
```

```python
df = pd.DataFrame(data)
```

```python
df
```

# Target Variable

```python
x = df.drop(["result"], axis = 1)
y = df[["result"]]

h = x
```

```python
x
```

```python
y
```

# train test split

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, random_state = 21)
```

# feature scaling

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
```

```python
trian_sc = sc.fit_transform(xtrain)
```

```python
test_sc = sc.transform(xtest)
```

# model

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(trian_sc, ytrain)
```

# predict

```python
y_predict = model.predict(test_sc)
```

```python
y_predict
```

# Evaluation the model

```python
from warnings import filterwarnings
filterwarnings("ignore")
from sklearn.metrics import accuracy_score, classification_report

acc = accuracy_score(ytest, y_predict)
```

```python
acc
```

```python
ytest
```

```python
classification_report(ytest, y_predict)
```

# hyperparameter selection of k

```python
for k in range(1, 9):
    knn = KNeighborsClassifier(n_neighbors = k)
    knn.fit(trian_sc, ytrain)
    y_pred = knn.predict(test_sc)
    acc = accuracy_score(ytest, y_pred)
    print(k, acc)
```
