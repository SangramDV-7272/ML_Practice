## What is boosting?

**Boosting** is an ensemble machine-learning technique that combines many weak models—usually small decision trees—to create one strong model.

A **weak learner** is a model that performs only slightly better than random guessing. Boosting trains these learners **sequentially**, where each new model focuses on correcting the errors made by the previous models.

### Basic idea

Suppose we want to predict house prices:

1. The first small tree makes predictions.
2. We calculate its errors.
3. The next tree focuses more on the incorrectly predicted houses.
4. We repeat this process many times.
5. The final prediction combines the outputs of all trees.

Conceptually:

```text
Final model =
Tree 1 + Tree 2 + Tree 3 + ... + Tree N
```

Each tree usually has a small contribution controlled by a **learning rate**.

---

## Boosting versus Random Forest

| Random Forest | Boosting |
|---|---|
| Trees are built independently | Trees are built sequentially |
| Can train trees in parallel | Later trees depend on earlier trees |
| Mainly reduces variance | Mainly reduces bias and improves errors |
| Usually less sensitive to parameters | Often needs careful tuning |
| Less likely to overfit with default settings | Can overfit if too many trees are used |

---

# Main boosting algorithms

## 1. AdaBoost

**AdaBoost**, or Adaptive Boosting, gives more importance to incorrectly predicted training samples.

### Process

1. Start with equal weights for all samples.
2. Train a weak model.
3. Increase the weights of incorrectly predicted samples.
4. Train the next model with these updated weights.
5. Combine all models using weighted voting or weighted prediction.

For classification, incorrectly classified observations receive greater weight.

For regression, a related method is called **AdaBoost.R2**.

```python
from sklearn.ensemble import AdaBoostRegressor

model = AdaBoostRegressor(
    n_estimators=100,
    learning_rate=0.05,
    random_state=42
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

### Advantages

- Simple and effective
- Works well with weak learners
- Can improve underperforming models

### Disadvantages

- Sensitive to noisy data and outliers
- A few bad observations can receive too much weight

---

## 2. Gradient Boosting

Gradient Boosting does not directly change sample weights. Instead, every new tree learns to predict the **residual errors** of the current model.

For regression:

```text
Residual = Actual value - Predicted value
```

### Example

Suppose:

```text
Actual price:    300,000
First prediction:250,000
Residual:         50,000
```

The next tree tries to learn these residuals and correct the first prediction.

```python
from sklearn.ensemble import GradientBoostingRegressor

model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

### Important parameters

- `n_estimators`: Number of trees
- `learning_rate`: Contribution of each tree
- `max_depth`: Maximum depth of each tree
- `subsample`: Fraction of samples used per tree

A common relationship is:

```text
More trees + smaller learning rate = often better generalization
```

However, training may take longer.

---

## 3. XGBoost

**XGBoost** means **Extreme Gradient Boosting**. It is an optimized and regularized version of gradient boosting.

It is popular because it:

- Trains efficiently
- Supports regularization
- Handles missing values
- Can work with large datasets
- Often performs very well on tabular data

Installation:

```bash
pip install xgboost
```

Regression example:

```python
from xgboost import XGBRegressor

model = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

Important XGBoost parameters:

- `n_estimators`: Number of boosting rounds
- `learning_rate`: Step size
- `max_depth`: Tree complexity
- `subsample`: Row sampling
- `colsample_bytree`: Feature sampling
- `reg_alpha`: L1 regularization
- `reg_lambda`: L2 regularization

---

**CatBoost** is designed to work especially well with categorical features.

It can process categories such as:

```text

Department = Sales
```

without requiring extensive manual one-hot encoding.

Installation:

```bash
pip install catboost
```

Example:

```python

model = CatBoostRegressor(

```

CatBoost is often a good choice when your dataset contains many categorical columns.

# How boosting works for regression

For regression, the model is built incrementally:

```python
initial_prediction = average(y_train)

for each tree:
    residual = actual_values - current_predictions
    train_tree_on(residual)
    current_predictions += learning_rate * tree_prediction
```

The final prediction is approximately:

```text
Prediction =
Initial prediction
+ learning rate × Tree 1
+ learning rate × Tree 2
+ ...
```

# Advantages of boosting

- Usually gives strong predictive performance
- Works well with tabular data
- Can model nonlinear relationships
- Can capture feature interactions
- Often does not require feature scaling
- Useful for both classification and regression

# Disadvantages of boosting

- Can overfit if not tuned carefully
- Sequential training can be slower
- Sensitive to noisy data and outliers
- Has several parameters to tune
- Less interpretable than a single decision tree

---

# Common ways to reduce overfitting

Use:

```python
n_estimators=100
learning_rate=0.05
max_depth=3
```

Also consider:

- Using a smaller `max_depth`
- Reducing `learning_rate`
- Using early stopping
- Increasing regularization
- Using a validation set
- Checking performance on unseen test data

## Which algorithm should you choose?

- **Simple scikit-learn solution:** `GradientBoostingRegressor`
- **High accuracy on tabular data:** `XGBRegressor`
- **Very large dataset:** `LGBMRegressor`
- **Many categorical features:** `CatBoostRegressor`
- **Simple adaptive boosting:** `AdaBoostRegressor`

## Q. Can we use accuracy_score in regration model?

No. **`accuracy_score` is generally not used for regression models.**

`accuracy_score` is designed for **classification**, where predictions are discrete classes such as:

```text
Cat, Dog
0, 1
```

Regression predicts continuous numerical values, such as:

```text
125.6, 240000.75, 32.4
```

A prediction of `100.1` instead of `100.0` is not exactly equal, so accuracy would not be meaningful.

## Metrics for regression

Use these instead:

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
```

Example:

```python
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print("R²:", r2)
print("MAE:", mae)
print("MSE:", mse)
```

You can also calculate RMSE:

```python
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("RMSE:", rmse)
```

### Meaning of the metrics

- **R²:** How well the model explains the variation in the target. Higher is generally better.
- **MAE:** Average absolute prediction error. Lower is better.
- **MSE:** Average squared prediction error. Lower is better.
- **RMSE:** Square root of MSE, expressed in the same units as the target. Lower is better.

If your target is converted into categories, such as `"low"`, `"medium"`, and `"high"`, then it becomes a **classification problem**, and `accuracy_score` may be appropriate.
