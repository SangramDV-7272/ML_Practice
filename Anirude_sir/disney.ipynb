```python
# rating

def rating(x):
    if x >= 4.0:
        return "Good"
    elif x >= 3.0:
        return "Average"
    else:
        return "Poor"
    
disney["rating_category"] = disney["rating"].apply(rating)
```

```python
disney["rating_category"].value_counts().plot(kind = "pie", autopct = '%1.1f%%')
```

```python
def watching_time(x):
    if x < 30.0:
        return "Short"
    elif x < 60.0:
        return "Medium"
    elif x < 120.0:
        return "Long"
    elif x < 180.0:
        return "Very Long"
    else:
        return "to much Long"
    
disney["watch_time_category"] = disney["watch_time_min"].apply(watching_time)
```
