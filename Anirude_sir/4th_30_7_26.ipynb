```python
def name(x):
    if x == 0:
        return "setosa"
    elif x == 1:
        return 'versicolor'
    else:
        return 'virginica'


df["target_name"] = df["target"].apply(name)
```

```python
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
```

```python
wine
```

```python
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
```

```python
wine_df["target"] = wine.target
```

```python
def wine_name(x):
    if x == 0:
        return 'class_0'
    elif x == 1:
        return 'class_1'
    else:
        return 'class_2'


wine_df["target_name"] = wine_df["target"].apply(wine_name)
```
